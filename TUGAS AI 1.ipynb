# 5 Prompt Geofisika dengan Struktur R–T–C–I–K–O–Q–V

## Struktur Prompt

| Kode | Komponen | Penjelasan |
|:---:|---|---|
| **R** | **Role** | Peran AI yang harus dijalankan |
| **T** | **Task** | Tugas yang harus dikerjakan |
| **C** | **Context** | Latar belakang atau konteks masalah |
| **I** | **Input** | Data yang diberikan |
| **K** | **Constraints** | Batasan atau ketentuan |
| **O** | **Output** | Hasil yang diharapkan |
| **Q** | **Quality Criteria** | Kriteria kualitas hasil |
| **V** | **Verification** | Verifikasi atau pengecekan hasil |

# Prompt 1 — Pemetaan Zona Intrusi Air Laut

### R — Role

Anda berperan sebagai **ahli geofisika eksplorasi air tanah dan hidrogeologi** yang berpengalaman dalam interpretasi data geolistrik.

### T — Task

Analisis data geolistrik untuk mengidentifikasi **zona yang diduga mengalami intrusi air laut** dan menentukan batas antara zona air tawar dan air asin berdasarkan distribusi nilai resistivitas bawah permukaan.

### C — Context

Suatu wilayah pesisir mengalami penurunan kualitas air tanah. Air sumur masyarakat mulai menunjukkan peningkatan salinitas. Survei geolistrik dilakukan untuk mengetahui distribusi resistivitas bawah permukaan dan mengidentifikasi kemungkinan penyebaran air laut ke dalam akuifer.

### I — Input

Data yang diberikan berupa:

- Koordinat titik pengukuran.
- Nilai resistivitas.
- Kedalaman investigasi.
- Penampang resistivitas 2D.
- Data geologi atau hidrogeologi jika tersedia.
- Informasi lokasi sumur masyarakat jika tersedia.

### K — Constraints

- Gunakan hanya data yang diberikan.
- Jangan mengarang nilai resistivitas atau kedalaman.
- Jangan menentukan zona intrusi hanya berdasarkan satu nilai resistivitas tanpa mempertimbangkan kondisi geologi.
- Jelaskan hubungan antara nilai resistivitas rendah dengan kemungkinan keberadaan air asin.
- Bedakan antara data hasil pengukuran dan interpretasi.
- Jika terdapat ketidakpastian, jelaskan sumber ketidakpastian tersebut.

### O — Output

Berikan:

1. Interpretasi distribusi resistivitas bawah permukaan.
2. Identifikasi zona yang diduga mengalami intrusi air laut.
3. Perkiraan kedalaman zona intrusi jika data memungkinkan.
4. Perbandingan antara zona air tawar dan zona yang diduga terintrusi.
5. Tabel hasil interpretasi.
6. Penjelasan geofisika.
7. Kesimpulan.

### Q — Quality Criteria

Jawaban harus berdasarkan prinsip geolistrik dan hidrogeologi, menggunakan data secara konsisten, mudah dipahami mahasiswa S1 Geofisika, dan membedakan hasil pengamatan dengan interpretasi.

### V — Verification

Periksa kembali kesesuaian antara **nilai resistivitas, kedalaman, posisi geografis, dan interpretasi zona intrusi** sebelum memberikan kesimpulan akhir.

# Prompt 2 — Investigasi Kondisi Fondasi Bangunan

### R — Role

Anda berperan sebagai **ahli geofisika teknik dan geoteknik** yang berpengalaman dalam investigasi kondisi bawah permukaan untuk pembangunan infrastruktur.

### T — Task

Analisis data geofisika untuk mengetahui **kondisi bawah permukaan dan kedalaman batuan dasar** sebagai bahan pertimbangan dalam pembangunan sebuah gedung.

### C — Context

Sebuah gedung bertingkat akan dibangun pada suatu wilayah yang belum diketahui kondisi bawah permukaannya secara detail. Investigasi geofisika diperlukan untuk mengetahui variasi lapisan tanah, kedalaman batuan dasar, dan kemungkinan adanya zona lemah yang dapat memengaruhi konstruksi.

### I — Input

Data yang diberikan dapat berupa:

- Data seismik refraksi.
- Kecepatan gelombang P.
- Data geolistrik.
- Penampang bawah permukaan.
- Data kedalaman.
- Informasi geologi lokasi.

### K — Constraints

- Jangan menentukan jenis material hanya berdasarkan satu parameter.
- Gunakan hubungan antara kecepatan gelombang, resistivitas, dan kondisi geologi.
- Jangan mengarang nilai parameter yang tidak tersedia.
- Bedakan antara zona batuan keras dan zona material lebih lemah.
- Jelaskan keterbatasan interpretasi geofisika.

### O — Output

Berikan:

1. Interpretasi lapisan bawah permukaan.
2. Perkiraan kedalaman batuan dasar.
3. Identifikasi zona material yang relatif lemah.
4. Interpretasi kondisi bawah permukaan untuk kebutuhan konstruksi.
5. Tabel hasil interpretasi.
6. Penjelasan dasar geofisika.
7. Kesimpulan.

### Q — Quality Criteria

Hasil analisis harus menggunakan prinsip geofisika teknik yang benar, sistematis, dan tidak memberikan kesimpulan yang melebihi kemampuan data.

### V — Verification

Periksa kembali **nilai kecepatan gelombang, resistivitas, kedalaman, dan hubungan antarparameter** untuk memastikan interpretasi konsisten.

# Prompt 3 — Deteksi Kebocoran Bendungan

### R — Role

Anda berperan sebagai **ahli geofisika teknik yang berfokus pada investigasi bendungan dan struktur bawah permukaan**.

### T — Task

Analisis data geofisika untuk mengidentifikasi **zona yang diduga mengalami rembesan atau kebocoran pada tubuh dan fondasi bendungan**.

### C — Context

Sebuah bendungan menunjukkan indikasi peningkatan rembesan air pada bagian hilir. Survei geofisika dilakukan untuk mengetahui kemungkinan adanya jalur rembesan bawah permukaan dan zona dengan kondisi material yang berbeda dari lingkungan sekitarnya.

### I — Input

Data yang diberikan dapat berupa:

- Penampang resistivitas.
- Data self-potential (SP).
- Data geolistrik.
- Posisi titik pengukuran.
- Elevasi.
- Informasi struktur bendungan.

### K — Constraints

- Jangan menyatakan adanya kebocoran secara pasti hanya berdasarkan satu anomali.
- Identifikasi terlebih dahulu karakteristik anomalinya.
- Jelaskan hubungan anomali dengan kemungkinan jalur rembesan.
- Jangan mengarang data yang tidak tersedia.
- Pertimbangkan kemungkinan penyebab lain dari anomali.

### O — Output

Berikan:

1. Identifikasi anomali geofisika.
2. Posisi anomali.
3. Perkiraan kedalaman anomali jika tersedia.
4. Interpretasi kemungkinan jalur rembesan.
5. Perbandingan dengan kondisi sekitar.
6. Tabel hasil analisis.
7. Kesimpulan.

### Q — Quality Criteria

Interpretasi harus membedakan dengan jelas antara **data, anomali, interpretasi, dan kesimpulan**, serta menggunakan prinsip geofisika teknik yang tepat.

### V — Verification

Bandingkan hasil dari beberapa parameter atau metode jika tersedia. Pastikan lokasi anomali konsisten dengan informasi kondisi bendungan dan lingkungan sekitarnya.

# Prompt 4 — Eksplorasi Mineralisasi

### R — Role

Anda berperan sebagai **ahli geofisika eksplorasi mineral** yang berpengalaman dalam interpretasi data geofisika untuk eksplorasi sumber daya mineral.

### T — Task

Analisis data geofisika untuk mengidentifikasi **zona yang memiliki karakteristik anomali dan berpotensi berkaitan dengan mineralisasi**.

### C — Context

Suatu wilayah menunjukkan indikasi mineralisasi berdasarkan kondisi geologi permukaan. Survei geofisika dilakukan untuk mengetahui pola anomali bawah permukaan dan menentukan area yang dapat dipertimbangkan sebagai target survei lanjutan.

### I — Input

Data yang diberikan dapat berupa:

- Data magnetik.
- Data gravitasi.
- Data IP atau induced polarization.
- Data resistivitas.
- Peta geologi.
- Koordinat titik pengukuran.

### K — Constraints

- Jangan menyatakan keberadaan mineral tertentu hanya berdasarkan satu metode geofisika.
- Gunakan kombinasi data geologi dan geofisika jika tersedia.
- Bedakan antara anomali geofisika dan interpretasi mineralisasi.
- Jangan mengarang nilai anomali.
- Jelaskan kemungkinan penyebab lain dari setiap anomali.

### O — Output

Berikan:

1. Identifikasi anomali geofisika.
2. Interpretasi karakteristik setiap anomali.
3. Hubungan anomali dengan kondisi geologi.
4. Identifikasi zona yang dapat menjadi target eksplorasi lanjutan.
5. Tabel perbandingan zona.
6. Tingkat keyakinan interpretasi berdasarkan data.
7. Kesimpulan.

### Q — Quality Criteria

Analisis harus mengintegrasikan informasi geologi dan geofisika serta membedakan secara jelas antara **hasil pengukuran, anomali, interpretasi, dan dugaan mineralisasi**.

### V — Verification

Periksa konsistensi antara **anomali magnetik, gravitasi, IP, resistivitas, dan kondisi geologi** sebelum menentukan target eksplorasi.

# Prompt 5 — Analisis Potensi Bahaya Gempa untuk Infrastruktur

### R — Role

Anda berperan sebagai **ahli seismologi dan geofisika teknik** yang berfokus pada kajian bahaya gempa.

### T — Task

Analisis data kegempaan dan kondisi bawah permukaan untuk mengidentifikasi **karakteristik wilayah yang perlu diperhatikan dalam perencanaan dan mitigasi bahaya gempa**.

### C — Context

Suatu kawasan berkembang menjadi wilayah permukiman dan pembangunan infrastruktur. Wilayah tersebut memiliki aktivitas gempa yang perlu dikaji untuk memahami pola kegempaan serta kondisi bawah permukaan yang dapat memengaruhi respons tanah terhadap getaran gempa.

### I — Input

Data yang diberikan berupa:

- Data katalog gempa.
- Magnitudo.
- Koordinat episentrum.
- Kedalaman hiposentrum.
- Data mikrotremor jika tersedia.
- Data kecepatan gelombang Vs jika tersedia.
- Peta geologi.

### K — Constraints

- Jangan memprediksi waktu terjadinya gempa.
- Jangan menyatakan suatu wilayah pasti mengalami gempa.
- Gunakan data historis sebagai dasar analisis.
- Bedakan antara aktivitas kegempaan dan tingkat bahaya.
- Jangan mengarang data yang tidak tersedia.
- Jelaskan keterbatasan data.

### O — Output

Berikan:

1. Analisis distribusi episentrum.
2. Analisis distribusi kedalaman gempa.
3. Identifikasi pola kegempaan.
4. Interpretasi kondisi bawah permukaan jika data Vs tersedia.
5. Hubungan antara kondisi geologi dan aktivitas gempa.
6. Identifikasi aspek yang perlu diperhatikan dalam mitigasi.
7. Kesimpulan.

### Q — Quality Criteria

Jawaban harus berdasarkan prinsip seismologi dan geofisika teknik, menggunakan data secara hati-hati, tidak melakukan prediksi gempa, serta mudah dipahami mahasiswa S1 Geofisika.

### V — Verification

Periksa kembali **koordinat, magnitudo, kedalaman, distribusi gempa, nilai Vs, dan informasi geologi** sebelum memberikan interpretasi akhir.
