## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

## 2. Recipe Normalisasi GeoJSON

### Problem

Susun recipe untuk mengubah file GeoJSON menjadi GeoJSON yang
dinormalisasi, inventaris CSV, dan ringkasan JSON. Recipe harus
memeriksa geometri, properti, geometri kosong, serta urutan koordinat
tanpa menebak informasi yang tidak tersedia.

### Minimal Theory

GeoJSON menggunakan struktur JSON untuk menyimpan fitur geografis.
Pemeriksaan perlu mencakup struktur `FeatureCollection`, tipe geometri,
koordinat, properti, dan sistem referensi koordinat. Untuk koordinat
geografis GeoJSON, urutan koordinat umumnya berupa bujur lalu lintang.

### Input dan Constraints

-   `GEOJSON_PATH`: path GeoJSON sumber.
-   Path output untuk GeoJSON, CSV inventaris, dan JSON ringkasan.
-   Jangan menghilangkan fitur atau properti tanpa aturan yang jelas.
-   Jangan menebak CRS atau mengubah urutan koordinat tanpa dasar.
-   Keberhasilan membaca file tidak otomatis membuktikan topologi
    geometri valid.

### Prompt dan Procedure

> Susun recipe untuk menormalisasi `[GEOJSON_PATH]` menjadi GeoJSON,
> inventaris CSV, dan ringkasan JSON. Periksa struktur, tipe geometri,
> properti, geometri kosong atau tidak valid, urutan koordinat, dan
> informasi CRS. Nyatakan ketidakpastian secara terbuka dan jangan
> mengklaim output telah dihasilkan sebelum eksekusi dibuktikan.

Langkah kerja: 1. Baca dan validasi sintaks JSON. 2. Periksa struktur
objek dan daftar fitur. 3. Periksa geometri, koordinat, serta properti.
4. Identifikasi CRS atau urutan koordinat yang belum jelas. 5.
Normalisasi hanya berdasarkan aturan yang terdefinisi. 6. Ekspor
inventaris fitur dan ringkasan hasil pemeriksaan. 7. Catat fitur yang
gagal diproses.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan pustaka yang mendukung
GeoJSON, misalnya `json` untuk pemeriksaan struktur dan pustaka
geospasial yang tersedia di lingkungan kerja. Dependency harus dicatat
dan tidak boleh diasumsikan sudah terpasang.

### Output

-   GeoJSON hasil normalisasi.
-   CSV inventaris fitur.
-   JSON ringkasan QA/QC.
-   Run record serta peringatan jika tersedia.

### Optional Check / Evidence

Bandingkan jumlah fitur sumber dan hasil, periksa properti penting,
validasi struktur output, dan catat fitur yang tidak berhasil diproses.
Jika program belum dijalankan, sebutkan output sebagai hasil yang
direncanakan.

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (\alpha)
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=\frac{\alpha\Delta t}{(\Delta x)^2} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r \le 1/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

## 2. Recipe QA/QC dan Visualisasi Data Gempa Bumi

### Problem

Membuat workflow yang dapat digunakan untuk memeriksa kualitas data gempa bumi sebelum dilakukan analisis seismologi. Pemeriksaan meliputi kelengkapan kolom, nilai koordinat, kedalaman, magnitudo, waktu kejadian, nilai yang hilang, serta data duplikat.

Recipe ini juga bertujuan menghasilkan ringkasan QA/QC dan visualisasi sebaran episenter berdasarkan koordinat geografis. Hasil pemeriksaan digunakan untuk mengetahui apakah data layak digunakan dalam analisis lanjutan, seperti pemetaan distribusi gempa, analisis kedalaman, dan identifikasi pola spasial kejadian gempa.

Fokus utama recipe adalah pemeriksaan kualitas data dan visualisasi dasar, bukan prediksi gempa bumi atau penentuan mekanisme sumber gempa.

### Minimal Theory

Data gempa bumi umumnya memuat informasi waktu kejadian, koordinat lintang dan bujur, kedalaman hiposenter, serta magnitudo. Informasi tersebut digunakan untuk menggambarkan distribusi spasial dan karakteristik dasar kejadian gempa bumi.

Koordinat geografis digunakan untuk menentukan posisi episenter pada permukaan bumi. Lintang menunjukkan posisi utara atau selatan terhadap ekuator, sedangkan bujur menunjukkan posisi timur atau barat terhadap meridian utama.

Kedalaman hiposenter menunjukkan jarak vertikal sumber gempa dari permukaan bumi, sedangkan magnitudo digunakan untuk menyatakan ukuran gempa berdasarkan skala pengukuran yang digunakan oleh sumber data.

Pemeriksaan kualitas data atau Quality Assurance/Quality Control (QA/QC) dilakukan untuk mengidentifikasi kesalahan struktur, nilai yang tidak masuk akal, nilai hilang, serta catatan kejadian yang terduplikasi. Sebagai contoh, lintang geografis harus berada pada rentang -90 hingga 90 derajat, sedangkan bujur harus berada pada rentang -180 hingga 180 derajat.

Data yang lolos pemeriksaan awal belum tentu sepenuhnya benar secara ilmiah. Validitasnya tetap bergantung pada sumber data, metode pengumpulan, ketelitian instrumen, dan metadata yang tersedia.

### Input dan Constraints

Input utama:

- `DATA_PATH`: lokasi file CSV data gempa bumi.
- `TIME_COLUMN`: nama kolom waktu kejadian gempa.
- `LAT_COLUMN`: nama kolom lintang episenter.
- `LON_COLUMN`: nama kolom bujur episenter.
- `DEPTH_COLUMN`: nama kolom kedalaman hiposenter.
- `MAG_COLUMN`: nama kolom magnitudo.

Input tambahan jika tersedia:

- `EVENT_ID`: identitas unik kejadian gempa.
- `MAG_TYPE`: jenis magnitudo, misalnya Mw, ML, atau mb.
- `SOURCE`: sumber data, misalnya katalog lembaga pemantauan gempa.
- `REGION`: nama wilayah kejadian gempa.

Batasan:

1. File sumber harus dipertahankan dan dibaca secara read-only.
2. Kolom wajib harus diperiksa sebelum analisis dilanjutkan.
3. Nilai lintang dan bujur di luar rentang geografis yang valid harus ditandai sebagai masalah QA/QC.
4. Nilai kedalaman atau magnitudo yang hilang harus dilaporkan tanpa mengisi nilai secara otomatis.
5. Timestamp duplikat tidak otomatis berarti kejadian gempa yang sama; pemeriksaan perlu mempertimbangkan ID kejadian, koordinat, waktu, dan metadata lainnya.
6. Data yang mencurigakan harus ditandai, bukan dihapus secara otomatis.
7. Visualisasi harus mencantumkan satuan, label sumbu, dan keterangan sumber data jika tersedia.
8. Hasil QA/QC tidak boleh dianggap sebagai bukti bahwa katalog gempa sepenuhnya akurat.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda bertindak sebagai asisten pengolahan data seismologi. Baca file `[DATA_PATH]` secara read-only dan lakukan pemeriksaan kualitas data gempa bumi.
>
> Periksa struktur kolom, tipe data, kelengkapan timestamp, nilai hilang, koordinat lintang dan bujur, kedalaman hiposenter, magnitudo, serta kemungkinan duplikasi kejadian. Gunakan nama kolom yang ditentukan dalam konfigurasi dan jangan mengasumsikan bahwa kolom memiliki nama tertentu jika belum diperiksa.
>
> Buat laporan QA/QC yang merangkum jumlah baris, jumlah kolom, kolom yang ditemukan, rentang waktu kejadian, nilai hilang, nilai di luar rentang yang diizinkan, dan kandidat data duplikat.
>
> Buat visualisasi sebaran episenter menggunakan koordinat lintang dan bujur, serta visualisasi tambahan hubungan magnitudo dan kedalaman apabila kolom yang diperlukan tersedia.
>
> Pisahkan data asli, hasil pemeriksaan, dan interpretasi. Jangan menghapus atau memperbaiki data sumber secara otomatis. Catat status eksekusi secara akurat dan jangan mengklaim bahwa program telah dijalankan apabila belum terdapat bukti eksekusi.

**Procedure:**

1. Baca file CSV secara read-only menggunakan pustaka pengolahan data.
2. Periksa struktur tabel dan keberadaan kolom yang diperlukan.
3. Validasi tipe data waktu, lintang, bujur, kedalaman, dan magnitudo.
4. Hitung jumlah nilai hilang pada setiap kolom.
5. Identifikasi koordinat di luar rentang geografis yang valid dan nilai numerik yang mencurigakan.
6. Periksa timestamp duplikat dan kandidat kejadian yang mungkin terduplikasi berdasarkan informasi yang tersedia.
7. Hitung rentang waktu, rentang magnitudo, dan rentang kedalaman untuk data yang dapat diperiksa.
8. Buat visualisasi sebaran episenter berdasarkan koordinat geografis.
9. Buat grafik magnitudo terhadap kedalaman apabila data yang diperlukan tersedia.
10. Simpan laporan QA/QC, grafik, dan catatan eksekusi tanpa mengubah file sumber.
11. Periksa kesesuaian jumlah data, kelengkapan laporan, dan keberhasilan penyimpanan output.

### Code / Implementation Plan

Implementasi direncanakan menggunakan Python dengan pustaka berikut:

- `pandas`: membaca CSV, memeriksa kolom, menghitung nilai hilang, dan mengidentifikasi kandidat duplikasi.
- `numpy`: melakukan pemeriksaan numerik dan perhitungan statistik dasar.
- `matplotlib`: membuat grafik sebaran episenter serta hubungan magnitudo dan kedalaman.
- `json`: menyimpan ringkasan konfigurasi dan catatan eksekusi.
- `pathlib`: mengelola path file input dan output.

Jika diperlukan, pustaka pemetaan geografis dapat ditambahkan untuk menghasilkan peta dengan garis pantai dan batas wilayah. Penggunaan pustaka tambahan harus disesuaikan dengan lingkungan Python yang tersedia.

Contoh rancangan struktur direktori proyek:

```text
earthquake-data-qc/
├── data/
│   └── earthquake-catalog.csv
├── code/
│   └── earthquake_qc.py
└── results/
    ├── earthquake-qc-report.md
    ├── epicenter-distribution.png
    ├── magnitude-depth.png
    └── run-record.json
```

Struktur direktori tersebut merupakan rancangan implementasi. Keberadaan file dan keberhasilan program harus diverifikasi setelah proses eksekusi dilakukan.

### Output

1. `earthquake-qc-report.md`: laporan QA/QC yang berisi jumlah baris dan kolom, kolom yang ditemukan, rentang waktu, jumlah nilai hilang, pemeriksaan rentang koordinat, statistik magnitudo dan kedalaman, serta jumlah kandidat duplikasi.

2. `epicenter-distribution.png`: visualisasi sebaran episenter berdasarkan lintang dan bujur. Grafik harus mencantumkan label sumbu, satuan koordinat, dan judul yang sesuai.

3. `magnitude-depth.png`: grafik hubungan magnitudo dengan kedalaman hiposenter jika data yang diperlukan tersedia dan valid.

4. `run-record.json`: catatan eksekusi yang mencakup waktu pelaksanaan, konfigurasi input, status pemeriksaan, jumlah data yang diperiksa, peringatan, dan status penyelesaian program.

5. `qc-findings.csv` (opsional): tabel temuan QA/QC yang mencatat baris atau identitas kejadian yang bermasalah, jenis masalah, dan keterangan pemeriksaan.

### Optional Check / Evidence

Checklist pemeriksaan:

- [ ] File CSV sumber tidak berubah selama proses audit.
- [ ] Kolom wajib berhasil diidentifikasi atau dilaporkan jika tidak tersedia.
- [ ] Tipe data waktu dan numerik diperiksa.
- [ ] Nilai hilang pada kolom penting berhasil dihitung.
- [ ] Koordinat di luar rentang geografis yang valid berhasil ditandai.
- [ ] Nilai kedalaman dan magnitudo yang mencurigakan dilaporkan berdasarkan aturan yang ditentukan.
- [ ] Timestamp duplikat dan kandidat kejadian duplikat diperiksa secara terpisah.
- [ ] Grafik episenter menggunakan koordinat yang valid dan label yang sesuai.
- [ ] Output yang direncanakan dibedakan dari file yang benar-benar dihasilkan.
- [ ] Status eksekusi dicatat secara akurat.

Status eksekusi yang diperbolehkan:

- `Draft`: spesifikasi recipe masih dalam tahap perancangan.
- `Not yet executed`: program belum dijalankan.
- `Partial`: sebagian pemeriksaan atau output telah berhasil diselesaikan.
- `Executed`: program telah dijalankan dan bukti yang diperlukan menunjukkan bahwa proses serta output yang dilaporkan berhasil diselesaikan.

### Keterbatasan dan Risiko

Hasil QA/QC sangat bergantung pada kelengkapan dan kualitas katalog gempa yang digunakan. Data dengan timestamp atau koordinat yang sama tidak selalu menunjukkan duplikasi karena beberapa kejadian dapat tercatat berdekatan dalam waktu dan ruang.

Perbedaan jenis magnitudo juga perlu diperhatikan. Magnitudo dari skala berbeda tidak boleh langsung dibandingkan tanpa mempertimbangkan definisi dan karakteristik skalanya.

Visualisasi sebaran episenter hanya menggambarkan distribusi kejadian berdasarkan data yang tersedia. Pola persebaran tersebut tidak secara otomatis membuktikan keberadaan sesar aktif, mekanisme tektonik tertentu, atau hubungan sebab-akibat antarperistiwa.

Interpretasi seismologi lanjutan perlu mempertimbangkan metadata katalog, ketidakpastian lokasi, kelengkapan deteksi gempa, serta informasi geologi dan tektonik regional.

### Ringkasan Recipe Baru

Recipe ini menyediakan kerangka kerja yang sistematis dan reproducible untuk memeriksa kualitas data gempa bumi sebelum digunakan dalam analisis seismologi. Proses dimulai dari validasi struktur tabel dan rentang nilai, pemeriksaan nilai hilang serta kandidat duplikasi, hingga visualisasi sebaran episenter dan hubungan magnitudo dengan kedalaman.

Dengan memisahkan data mentah, temuan QA/QC, hasil visualisasi, dan interpretasi ilmiah, recipe ini membantu meningkatkan transparansi pengolahan data serta mengurangi risiko kesalahan pada tahap analisis lanjutan. Seluruh hasil harus didasarkan pada data dan bukti eksekusi yang benar-benar tersedia.
